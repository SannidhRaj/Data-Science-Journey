```python
# Days 01 02 03 04 05  and their temperature values are given below. Create a line plot to visualize the temperature trend over the days.
import matplotlib.pyplot as plt
days=[1,2,3,4,5]
temperatures=[25,28,30,27,26]
plt.plot(days, temperatures,marker='o', linestyle='-', color='r',label='Temperature')
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Trend Over Days")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#basic syntax
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

days=np.array([1, 2, 3, 4, 5])
temperatures=np.array([25, 28, 30, 27, 26])
# Create a DataFrame from the data
df = pd.DataFrame({'Days': days, 'Temperature': temperatures})
sns.barplot(x='Days', y='Temperature', data=df,palette='deep')
sns.set_style("whitegrid")

plt.title("Temperature Trend Over Days")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.show()
```
