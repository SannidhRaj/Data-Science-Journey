### **1. BASIC SYNTAX**

```python
import requests

url = "https://example.com"
response = requests.get(url)

print(response.status_code)
print(response.text)
```

---

### **2. WITH HEADERS + TIMEOUT**

```python
import requests

url = "https://example.com"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(
    url,
    headers=headers,
    timeout=5
)

print(response.status_code)
print(response.text)
```

---

### **3. WITH ERROR HANDLING**

```python
import requests

url = "https://example.com"

headers = {
    "User-Agent": "Mozilla/5.0"
}

try:
    response = requests.get(
        url,
        headers=headers,
        timeout=5
    )

    response.raise_for_status()

    print(response.status_code)
    print(response.text)

except requests.exceptions.RequestException as e:
    print("Error:", e)
```
---
