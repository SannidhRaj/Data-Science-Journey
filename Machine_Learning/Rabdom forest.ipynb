`we have to predict will person servive or not`

```python

import pandas as pandas
import matplotlib.pyplot as plt 
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.tree import plot_tree

from sklearn.ensemble import BaggingClassifier
from sklearn.linear_model import LogisticRegression
```

```python
tit=sns.load_dataset("titanic")
```

```python
feature=["pclass","sex","fare","embarked","age"]
target=["survived"]

# Missing data 
from sklearn.impute import SimpleImputer
imp_median=SimpleImputer(strategy="median")
tit[["age"]]=imp_median.fit_transform(tit[["age"]])
imp_fre=SimpleImputer(strategy="most_frequent")
tit[["embarked"]]=imp_fre.fit_transform(tit[["embarked"]])
tit.isnull().sum()
tit.shape # out of 891 , 688 deck values are missing so we drop the deck column

#Encode
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
tit["sex"]=le.fit_transform(tit["sex"])
tit["embarked"]=le.fit_transform(tit["embarked"])

#feature
X=tit[feature]
y=tit["survived"]

#Train test split
X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42   
)
```

```python
# Decision Tree model( no pruning)

from sklearn.tree import DecisionTreeClassifier

model=DecisionTreeClassifier(
    max_depth=4
)
model.fit(X_train,y_train)
```

```python
#Prediction
from sklearn.metrics import accuracy_score

y_pred_test=model.predict(X_test)
y_pred_train=model.predict(X_train)
print(f" Training ACCURACY SCORE {accuracy_score(y_train,y_pred_train)}")
print(f"testing ACCURACY SCORE {accuracy_score(y_test,y_pred_test)}")
```

 Overfitting case ,
 we have to fix with random forest

```python
# To plot decision tree (NO PRUNING)

from sklearn.tree import plot_tree

plt.figure(figsize=(18,10))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Died","Survived"],
    filled=True,

)
plt.tight_layout
plt.show()
```

`Implimenting Random Forest`

```python
from sklearn.ensemble import RandomForestClassifier

model=RandomForestClassifier(
    n_estimators=501,
    max_depth=4,
    oob_score=True
)
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
print("this is more stable model")
print("Testing Accuracy: ",accuracy_score(y_test,y_pred))
print("OOB SCORE(Out of bag score ): ",model.oob_score_)
#this is more stable model
```
