```python
import numpy as np
```

```python
# Linear Regression

class LinearRegression:
    def __init__(self,learning_rate=0.1,n_iter=1000):
        self.bias=None
        self.weights=None
        self.lr=learning_rate
        self.n_iter=n_iter
        
    #model 
    def fit(self,X,y):
        n,m=X.shape # return rows and column (jitna column utna feature)
        self.bias=0
        self.weights=np.zeros(n)# it return many 0[0,0,0,0,.......]
        
        for i in range(self.n_iter): # Gradient Descent
            y_pred=self.bias+np.dot(X,self.weights) 
            
             #calculate derivative
            db=(1/m)*np.sum(y_pred-y)
            dw=(1/m)*np.dot(X.T(y_pred-y))
            
            #applying Convergence Theorem
            self.bias=self.bias-self.lr * db
            self.weights=self.weights-self.lr*dw
          
    def predict(self,X):
        y_pred=self.bias+np.dot(X,self.weights)
        return y_pred
    
        
        
        
        
        
```

```python
X=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

model=LinearRegression()
model.fit(X,y)

y_pred=model.predict(X)
print(y_pred)
```
